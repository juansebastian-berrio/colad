El entrenamiento de redes profundas presenta desafíos como el desvanecimiento de los gradientes, que puede ocurrir al utilizar funciones de activación como la sigmoide, especialmente en sus extremos, donde las curvas se vuelven planas y dificultan el aprendizaje.

Para solucionar estos problemas surgieron funciones de activación modernas, como ELU y SELU, que ayudan a mantener los valores de las activaciones controlados y permiten la auto-normalización. También destacan GELU, Swish y Mish, utilizadas en arquitecturas modernas por sus propiedades para mejorar el flujo del aprendizaje.

Otro elemento importante es la normalización, que busca centrar y escalar los datos o activaciones para facilitar y estabilizar el entrenamiento de las redes.

Finalmente, el aprendizaje por transferencia (transfer learning) permite reutilizar el conocimiento aprendido por una red en una tarea para resolver otra relacionada. Esta técnica contribuyó al resurgimiento y gran desarrollo de las redes neuronales profundas.
