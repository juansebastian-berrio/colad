Los modelos SVM (Support Vector Machine) son algoritmos de aprendizaje automático que se usan principalmente para clasificar datos, aunque también pueden utilizarse para problemas de regresión. La idea principal de SVM es encontrar una frontera que permita separar las diferentes clases de la mejor manera posible, tratando de dejar el mayor margen entre ellas.

Cuando los datos no se pueden separar fácilmente mediante una línea recta, SVM puede utilizar el truco del kernel para trabajar con relaciones más complejas. Uno de los kernels más utilizados es el RBF o gaussiano, que básicamente permite medir qué tan relacionados o cercanos están los diferentes puntos. En este modelo también son importantes los parámetros C y gamma. El parámetro C controla cuánto se penalizan los errores, mientras que gamma define qué tanta influencia tienen los puntos que están más cerca.

Además, existe SVR, que es la versión de SVM utilizada para problemas de regresión. En términos matemáticos, SVM busca ajustar los pesos para encontrar una frontera que separe las clases de la forma más adecuada. Por otro lado, SGDClassifier utiliza el descenso de gradiente estocástico, lo que permite ir ajustando los pesos del clasificador poco a poco a medida que procesa los datos.
