1. K-Vecinos más cercanos (KNN)
Clasifica un dato nuevo según los datos que tiene más cerca. Por ejemplo, puede determinar si una fruta es una manzana o una naranja comparándola con ejemplos conocidos.
2. Árbol de decisión
Toma decisiones mediante una estructura parecida a un árbol, haciendo preguntas o evaluaciones paso a paso hasta llegar a una respuesta.
3. Regresión lineal
Busca una relación entre variables y permite predecir un valor numérico. Por ejemplo, predecir el precio de una casa según su tamaño.
4. Regresión logística
Se utiliza principalmente para clasificación. Calcula la probabilidad de que un dato pertenezca a una categoría, como sí/no o spam/no spam.
5. Q-learning
Es un algoritmo de aprendizaje por refuerzo. Aprende qué acciones son mejores mediante recompensas y castigos hasta encontrar una estrategia adecuada.
6. Aprendizaje por refuerzo
Un agente aprende interactuando con un entorno: realiza acciones y recibe recompensas o penalizaciones para mejorar sus decisiones.
7. AdaBoost
Combina varios modelos sencillos para formar un modelo más potente. Da mayor importancia a los ejemplos que fueron clasificados incorrectamente.
8. Algoritmo genético
Está inspirado en la evolución biológica. Genera diferentes soluciones y las combina o modifica para encontrar una solución cada vez mejor a un problema.
9. Clasificador de Bayes ingenuo (Naive Bayes)
Utiliza probabilidades para clasificar datos. Es muy utilizado para tareas como clasificación de textos, detección de spam y análisis de documentos.
