El aprendizaje automático es un pilar fundamental de la inteligencia artificial que permite a los sistemas informáticos aprender a partir de datos para realizar predicciones, clasificaciones y tomar decisiones. Su funcionamiento se basa en suministrar información a algoritmos capaces de identificar patrones y construir modelos predictivos, dividiéndose principalmente en enfoques supervisados, no supervisados y por refuerzo.

Presente en sectores clave como la medicina, las finanzas, la educación y la tecnología, esta disciplina también enfrenta importantes desafíos relacionados con la calidad de los datos, el sobreajuste, los sesgos y la privacidad. Por lo tanto, comprender el aprendizaje automático no se limita a conocer sus algoritmos, sino también a entender sus profundas implicaciones éticas, sociales y prácticas.
