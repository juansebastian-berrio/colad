Los árboles de decisión son modelos de caja blanca porque se puede entender de forma sencilla cómo llegan a una respuesta. Se pueden usar para hacer predicciones y calcular probabilidades, y uno de los algoritmos más conocidos es CART, que utiliza criterios como Gini y entropía para decidir cómo dividir los datos. También existen los árboles de regresión, que sirven cuando queremos predecir valores numéricos. Para evitar que el árbol se vuelva demasiado grande y termine aprendiendo demasiado los datos de entrenamiento, se usa la regularización, que pone ciertos límites al modelo y ayuda a que funcione mejor con datos nuevos.
