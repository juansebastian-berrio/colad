Las redes neuronales con Keras se basan en la idea de imitar, de forma simplificada, el funcionamiento del cerebro humano, donde las neuronas se conectan entre sí para procesar información. Esta idea se llevó a la computación mediante el perceptrón, que es una de las piezas fundamentales de las redes neuronales artificiales. A partir de él surgieron modelos más complejos como el perceptrón multicapa (MLP), formado por diferentes capas de neuronas. Para entrenar estos modelos se utiliza principalmente la retropropagación, que permite ajustar los pesos de las conexiones para reducir los errores de las predicciones. En la práctica, estos modelos pueden construirse fácilmente con Keras, utilizando diferentes formas de crear redes, como la API Secuencial, la API Funcional y la API de Subclases, dependiendo de la complejidad y estructura que necesite el modelo.
