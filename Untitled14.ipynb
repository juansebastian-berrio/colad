La visión por computadora es un área de la inteligencia artificial inspirada en el funcionamiento biológico de la corteza visual, que permite a las máquinas analizar e interpretar imágenes. Las CNN (redes neuronales convolucionales) utilizan campos receptivos locales, capas y filtros para detectar características como bordes, formas y patrones. También utilizan la agrupación o pooling para reducir el tamaño de la información manteniendo las características importantes. A lo largo del tiempo han evolucionado diferentes arquitecturas de CNN, algunas muy conocidas y utilizadas para tareas como la clasificación de imágenes mediante modelos previamente entrenados y la detección de objetos, que permite identificar qué elementos aparecen en una imagen y dónde se encuentran.
